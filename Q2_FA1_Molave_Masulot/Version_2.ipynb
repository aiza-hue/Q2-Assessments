```python
# Function to add two numbers
def add_numbers(a, b):
    return a + b

# Function to subtract two numbers
def subtract_numbers(a, b):
    return a - b

# Function to multiply two numbers
def multiply_numbers(a, b):
    return a * b

# Function to divide two numbers
def divide_numbers(a, b):
    return a / b


# --- MAIN PROGRAM ---

# Step 1: Ask the user to enter two numbers
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Step 2: Show the menu choices
print("\nChoose operation:")
print("1 - Addition")
print("2 - Subtraction")
print("3 - Multiplication")
print("4 - Division")

# Step 3: Get user's choice
choice = input("\nEnter choice (1-4): ")

# Step 4: Perform calculation based on choice
if choice == "1":
    result = add_numbers(num1, num2)
    print("Result:", result)

elif choice == "2":
    result = subtract_numbers(num1, num2)
    print("Result:", result)

elif choice == "3":
    result = multiply_numbers(num1, num2)
    print("Result:", result)

elif choice == "4":
    # Check for division by zero before calling function
    if num2 == 0:
        print("Cannot divide by zero.")
    else:
        result = divide_numbers(num1, num2)
        print("Result:", result)

else:
    print("Invalid choice. Please pick 1, 2, 3, or 4.")
```

- ​Each function used two parameters named a and b to receive the numbers that need to be processed.
#### What arguments were passed when the functions were called?
- The variables num1 and num2 (which held the user's input values) were passed as arguments when calling the functions, such as add_numbers(num1, num2).
####How did your program use the returned value?
- ​The program stored the returned calculation in a variable called result, which was then printed out as the final answer.
